```python
import pandas as pd
import numpy as np
import chardet
```

```python
# 1) Detectar encoding para CSV
def detect_encoding(path, sample_bytes=50000):
    with open(path, "rb") as f:
        return chardet.detect(f.read(sample_bytes))["encoding"]
```

```python
# 4) Leer y perfilar cada archivo
for path in archivos:
    print(f"\nProcesando: {path}")
    try:
        if path.lower().endswith(".csv"):
            enc = detect_encoding(path) or "utf-8"
            print(f"Encoding detectado: {enc}")
            # on_bad_lines='skip' evita que una fila corrupta frene el análisis inicial
            df = pd.read_csv(path, encoding=enc, on_bad_lines="skip")
        else:
            # requiere openpyxl para .xlsx
            df = pd.read_excel(path)
        quick_report(df, os.path.basename(path))
    except Exception as e:
        print(f"⚠️ Error leyendo {path}: {e}")
        # Hint útil si se traba por separador
        if path.lower().endswith(".csv"):
            print("Tip: probá forzar separador con sep=',' o sep=';' y/o engine='python'.")
```
