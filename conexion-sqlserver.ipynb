```python
#Prueba de conexion y carga de esquemas para la separacion de capas
import pyodbc
import os
from dotenv import load_dotenv
```

```python
# Cargar variables de entorno desde el archivo .env
load_dotenv()

# Obtener variables de entorno y necesarias para la conexión
server = os.getenv("MSSQL_SERVER")
database = os.getenv("MSSQL_DATABASE")
username = os.getenv("MSSQL_USUARIO")
password = os.getenv("MSSQL_PASSWORD")
driver = os.getenv("MSSQL_DRIVER")
```

```python
# Cadena de conexión
connection_string = (
    f"DRIVER={{{driver}}};"
    f"SERVER={server};"
    f"DATABASE={database};"
    f"UID={username};"
    f"PWD={password}"
)
```

```python
try:
    conn = pyodbc.connect(connection_string)
    cursor = conn.cursor()
    cursor.execute("SELECT 1")
    print("Conectado correctamente", cursor.fetchone())
    conn.close()
except Exception as e:
    print("Error en la conexión:", e)
```

```python
# Creamos los esquemas Bronce, Silver y Gold
esquemas = ["Bronce", "Silver", "Gold"]
#POR LAS DUDAS ponemos IF NOT EXISTS pero como es nueva no tiene que existir
try:
    conn = pyodbc.connect(connection_string, autocommit=True)
    cursor = conn.cursor()
    for esquema in esquemas:
        cursor.execute(f"""
        IF NOT EXISTS (SELECT * FROM sys.schemas WHERE name = '{esquema}')
        BEGIN
            EXEC('CREATE SCHEMA {esquema}')
        END
        """)
    print("Esquemas creados con éxito")
    conn.close()
except Exception as e:
    print("Error al crear esquemas:", e)
```
