```python
import pandas as pd
import chardet
import os
```

```python
# 1) Detectar encoding para CSV
def detect_encoding(path, sample_bytes=50000):
    with open(path, "rb") as f:
        return chardet.detect(f.read(sample_bytes))["encoding"]
```

```python
def quick_report(df: pd.DataFrame, name: str):
    print(f"\n=== {name} ===")
    print(f"Filas: {df.shape[0]:,} | Columnas: {df.shape[1]}")
    print("\nTipos de datos:")
    print(df.dtypes)
    print("\nNulos:")
    print(df.isna().sum().sort_values(ascending=False).head(10))
    print("\nValores únicos (primeras 10 columnas):")
    for col in df.columns[:10]:
        print(f" - {col}: {df[col].nunique(dropna=True)} únicos")
    display(df.head(3))
```

```python
archivos = [
    "../Data/Bronce/2023.1_Sysarmy_Encuesta de remuneracin salarial Argentina.csv",
    "../Data/Bronce/2023.2_Sysarmy_Encuesta de remuneracin salarial Argentina.csv",
    "../Data/Bronce/Datos-Limpios-Sueldos2021.xlsx",
    "../Data/Bronce/indice-salarios-base-2012-valores-mensuales.csv",
]
```

```python
for path in archivos:
    print(f"\nProcesando: {path}")
    if path.lower().endswith(".csv"):
        enc = detect_encoding(path)
        print(f"Encoding detectado: {enc}")
        df = pd.read_csv(path, encoding=enc)
    else:
        df = pd.read_excel(path)
    quick_report(df, path)
```
