```python
def limpiar_provincia(df, col="provincia"):
    df = df.copy()
    # Unificamos CABA y PVBA en Buenos Aires
    df[col] = df[col].replace({
        "Ciudad Autónoma de Buenos Aires": "Buenos Aires",
        "Provincia de Buenos Aires": "Buenos Aires"
    })

    #mapeamos a ids
    mapping = {
        "Buenos Aires": 1,
        "Córdoba": 2,
        "Río Negro": 3,
        "Mendoza": 4,
        "San Luis": 5,
        "Jujuy": 6,
        "Entre Ríos": 7,
        "Tierra del Fuego": 8,
        "Santa Fe": 9,
        "Chaco": 10,
        "San Juan": 11,
        "Chubut": 12,
        "Corrientes": 13,
        "Misiones": 14,
        "Neuquén": 15,
        "Salta": 16,
        "Tucumán": 17,
        "La Rioja": 18,
        "Santa Cruz": 19,
        "La Pampa": 20,
        "Formosa": 21,
        "Santiago del Estero": 22,
        "Catamarca": 23
    }

    df[col] = df[col].map(mapping)
    return df, mapping
```

```python
def limpiar_jornada(df, col="jornada"):
    df = df.copy()
    mapping = {"Full-Time": 1, "Part-Time": 2}
    df[col] = df[col].map(mapping)
    return df, mapping
```

```python
def limpiar_tipo_contrato(df, col="tipo_contrato"):
    df = df.copy()
    mapping = {
    "Staff (planta permanente)": 1,
    "Remoto (empresa de otro país)": 2,
    "Tercerizado (trabajo a través de consultora o agencia)": 3,
    "Freelance": 4,
    "Participación societaria en una cooperativa": 5,
    "Contractor": 6
}
    df[col] = df[col].map(mapping)
    return df, mapping
```

```python
def limpiar_salario(df, col="salario_neto_ars"):
    df = df.copy()
    df[col] = pd.to_numeric(df[col], errors="coerce")
    return df
```

```python
def limpiar_sueldo_dolarizado(df, col="sueldo_dolarizado"):
    df = df.copy()

    def normalizar(x):
        if pd.isna(x):
            return np.nan
        texto = str(x).strip().lower()

        # Si dolarizado
        if texto in ["si", "sí", "true", "1"]:
            return 1
        if "usd" in texto or "dólar" in texto or "dolar" in texto:
            return 1  #para evitar perder valores del dolar buscamos palabras o frases que contengan para setearlo como 1

        # No dolarizado
        if texto in ["no", "false", "2"]:
            return 2
        if "no" in texto or "ninguno" in texto:
            return 2  #Por si escribieron no cobro en dolares o cosas así

        return np.nan  # valores no reconocidos quedan como nulos(usar la posible solucion)
    df[col] = df[col].apply(normalizar)
    mapping = {"Sí dolarizado": 1, "No dolarizado": 2}
    return df, mapping
```

```python
def limpiar_valor_dolar(df, col="valor_dolar"):
    df = df.copy()
    
    # Solo registros dolarizados osea que tienen marcado Si o devuelven 1
    mask_dolarizado = df['sueldo_dolarizado'] == 1
    
    def parse_dolar(x):
        try:
            if pd.isna(x) or x == "":
                return np.nan
            
            import re
            texto = str(x)
            
            # Buscar números en el texto (incluyendo decimales) para evitar perder datos útiles ya que algunos pusieron blue 1200 o parecido
            numeros = re.findall(r'\d+[.,]?\d*', texto)
            
            if numeros:
                # Tomar el primer número encontrado
                numero_str = numeros[0].replace(',', '.')
                valor = float(numero_str)
                
                # configuramos un rango valido en argentina para el valor dolar
                if 50 <= valor <= 1400:
                    return valor
            
            return np.nan
        except:
            return np.nan
    
    #Aplicamos el parseo de dolar para LOS QUE CUMPLAN EL FILTRO DE Q PUSIERON Q SI
    df.loc[mask_dolarizado, col] = df.loc[mask_dolarizado, col].apply(parse_dolar)
    
    # Dejamos los no como nulos
    df.loc[~mask_dolarizado, col] = np.nan
    
    return df
```

```python
def limpiar_edad(df, col="edad"):
    df = df.copy()
    
    def parse_edad(x):
        try:
            if pd.isna(x):
                return np.nan
            
            # Si viene como rango ejemplo: "30-40", tomar el primer valor y listo(explicar porque)
            edad_str = str(x).split("-")[0]
            edad = float(edad_str)
            
            # Validamos un rango para no tener datos de +80 años o menores de 16(por las dudas)
            if 16 <= edad <= 80:
                return edad
            
            return np.nan
        except:
            return np.nan
    
    df[col] = df[col].apply(parse_edad)
    return df
```

```python
def limpiar_genero(df, col="genero"):
    df = df.copy()
    mapping = {"Hombre": 1, "Mujer": 2, "No especificado": 3}
    def map_genero(x):
        val = unidecode.unidecode(str(x)).lower()
        if "hombre" in val or "varon" in val or "masculino" in val:
            return mapping["Hombre"]
        elif "mujer" in val or "femenino" in val:
            return mapping["Mujer"]
        else:
            return mapping["No especificado"]
    df[col] = df[col].apply(map_genero)
    return df, mapping
```

```python
def limpiar_bono(df, col="bono"):
    df = df.copy()
    mapping = {"Si": 1, "No": 2}
    df[col] = df[col].apply(lambda x: 1 if "no" not in str(x).lower() else 2)
    return df, mapping
```

```python
def limpiar_seniority(df, col="seniority"):
    df = df.copy()
    mapping = {"Junior": 1, "Semi-Senior": 2, "Senior": 3}
    df[col] = df[col].map(mapping)
    return df, mapping
```

```python

def limpiar_actualizacion_salarial(df, col="actualizacion_salarial"):
    df = df.copy()
    mapping = {"Si": 1, "No": 2}
    df[col] = df[col].apply(lambda x: 1 if "no" not in str(x).lower() else 2)
    return df, mapping
```

```python
def limpiar_satisfaccion_ingresos(df, col="satisfaccion_ingresos"):
    df = df.copy()
    
    # Mapeo más amplio basado en números ( ya vienen como 1,2,3,4)
    def map_satisfaccion(x):
        try:
            val = int(float(x))
            if 1 <= val <= 4:
                return val
            return 4  # Default a "Muy conforme" si está fuera de rango
        except:
            return 4  # Default si no se puede convertir
    
    df[col] = df[col].apply(map_satisfaccion)
    
    mapping = { "Muy disconforme": 1, "Disconforme": 2, "Conforme": 3, "Muy conforme": 4}
    return df, mapping
```

```python
def limpiar_buscando_trabajo(df, col="buscando_trabajo"):
    df = df.copy()
        
    mapping = {
        "Sí, estoy buscando activamente.": 1,
        "No, estoy muy conforme.": 2,
        "No, pero escucho ofertas.": 3
    }
    
    # Función de mapeo más amplia
    def map_buscando(x):
        if pd.isna(x):
            return 3
        
        x_str = str(x).lower()
        if "buscando activamente" in x_str or "sí" in x_str:
            return 1
        elif "muy conforme" in x_str:
            return 2
        else:
            return 3
    
    df[col] = df[col].apply(map_buscando)
    return df, mapping
```

```python
def mapear_puesto(df, col="puesto"):
    df = df.copy()
    mapping = {
        "Desarrollador": 1,
        "Data": 2,
        "QA": 3,
        "Infraestructura": 4,
        "Management": 5,
        "RRHH": 6,
        "Otros": 7
    }

    grupos = {
        "Developer": mapping["Desarrollador"],
        "Fullstack": mapping["Desarrollador"],
        "Technical Leader": mapping["Management"],
        "Product Manager": mapping["Management"],
        "Project Manager": mapping["Management"],
        "Manager / Director": mapping["Management"],
        "Recruiter / HR": mapping["RRHH"],
        "QA / Tester": mapping["QA"],
        "Sysadmin / Devops / Sre": mapping["Infraestructura"],
        "Data Engineer": mapping["Data"],
        "Data Scientist": mapping["Data"],
        "Bi Analyst / Data Analyst": mapping["Data"]
    }

    def map_func(x):
        return grupos.get(str(x), mapping["Otros"])

    df[col] = df[col].apply(map_func)
    return df, mapping
```

```python
def limpiar_experiencia_antiguedad(df):
    df = df.copy()
    for col in ["anios_experiencia", "antiguedad_empresa", "tiempo_puesto"]:
        df[col] = pd.to_numeric(df[col], errors="coerce")
        df[col] = df[col].fillna(0).round().astype(int)
    return df
```

```python
def limpiar_modalidad_dias(df, col="modalidad"):
    
    df = df.copy()
    mapping = {"Remoto": 1, "Hibrido": 2, "Presencial": 3}
    
    def map_modalidad(x):
        val = unidecode.unidecode(str(x)).lower()
        
        # PRIMERO híbrido (más específico) ya si ponemos remoto hay 2 opciones que contienen esa palabra
        if "hibrid" in val:
            return mapping["Hibrido"]
        elif "remoto" in val:
            return mapping["Remoto"]
        elif "presencial" in val:
            return mapping["Presencial"]
        else:
            return np.nan
    
    df[col] = df[col].apply(map_modalidad)
    
    # Ajustamos días presenciales según modalidad
    df.loc[df[col] == mapping["Remoto"], "dias_presenciales"] = 0
    df.loc[df[col] == mapping["Presencial"], "dias_presenciales"] = 5
    
    return df, mapping
```

```python
def limpiar_nivel_estudios(df, col="nivel_estudios"):
    df = df.copy()
    mapping = {
        "Secundario": 1,
        "Terciario": 2,
        "Universitario": 3,
        "Posgrado/Especialización": 4,
        "Maestría": 5,
        "Doctorado": 6,
        "Posdoctorado": 7,
        "No especificado": 8
    }
    
    
    # Convertir None/NaN a "No especificado"
    df[col] = df[col].fillna("No especificado")
    
    # Función para mapeo seguro
    def map_estudios(x):
        if x in mapping:
            return mapping[x]
        
        # Buscar por similitud limpiando los acentos y pasando a minúsculas, forma más facil de filtrar sin perder datos
        x_clean = unidecode.unidecode(str(x)).lower()
        for key, value in mapping.items():
            key_clean = unidecode.unidecode(key).lower()
            if key_clean in x_clean or x_clean in key_clean:
                return value
        
        return mapping["No especificado"]
    
    df[col] = df[col].apply(map_estudios)
    
    return df, mapping
```

```python
def limpiar_guardias(df, col="guardias"):
    df = df.copy()
    mapping = {"No": 3, "Si, Pasiva": 1, "Si, Activa": 2}
    
    def map_guardias(x):
        if x is None or str(x).lower() == "no":
            return mapping["No"]
        elif "pasiva" in str(x).lower():
            return mapping["Si, Pasiva"]
        elif "activa" in str(x).lower():
            return mapping["Si, Activa"]
        return mapping["No"]
    
    df[col] = df[col].apply(map_guardias)
    return df, mapping
```

```python
def verificar_y_limpiar_nans_final(df):
    #Limpieza final: elimina registros con NaN en columnas críticas. nos devuelve el DataFrame limpio esto para que no se filtren datos.
    # Columnas críticas que NO pueden tener NaN
    columnas_criticas = ['salario_neto_ars', 'edad']
    #Eliminamos los nulos de las columnas criticas
    df = df.dropna(subset=columnas_criticas)
    return df
```

```python
def limpiar_silver(df):
    mappings = {}
    #los que tienen numericos es porque no necesitan mapping
    df, mappings["provincia"] = limpiar_provincia(df)
    df, mappings["jornada"] = limpiar_jornada(df)
    df, mappings["tipo_contrato"] = limpiar_tipo_contrato(df)
    df = limpiar_salario(df)
    df, mappings["sueldo_dolarizado"] = limpiar_sueldo_dolarizado(df)
    df = limpiar_valor_dolar(df)
    df = limpiar_edad(df) 
    df, mappings["genero"] = limpiar_genero(df)
    df, mappings["seniority"] = limpiar_seniority(df) 
    df, mappings["bono"] = limpiar_bono(df)
    df, mappings["actualizacion_salarial"] = limpiar_actualizacion_salarial(df)
    df, mappings["satisfaccion_ingresos"] = limpiar_satisfaccion_ingresos(df)
    df, mappings["buscando_trabajo"] = limpiar_buscando_trabajo(df)
    df, mappings["puesto"] = mapear_puesto(df)
    df = limpiar_experiencia_antiguedad(df)
    df, mappings["modalidad"] = limpiar_modalidad_dias(df)
    df, mappings["nivel_estudios"] = limpiar_nivel_estudios(df)
    df, mappings["guardias"] = limpiar_guardias(df)
    df = verificar_y_limpiar_nans_final(df)

    return df, mappings
```

```python

def crear_tabla_genero(mapping, conn, schema="Silver"):
    cursor = conn.cursor()

    #con cursor ejecutamos el comando para crear la tabla e insertamos los valores del mapping que devuelve la funcioN
    cursor.execute(f"""
    CREATE TABLE {schema}.TBL_GENERO (
        ID INT PRIMARY KEY,
        DESCRIPCION NVARCHAR(100) NOT NULL
    );
    """)

    #Insertamos los valores del mapping que devuelve la funcion
    for desc, id_val in mapping.items():
        cursor.execute(
            f"INSERT INTO {schema}.TBL_GENERO (ID, DESCRIPCION) VALUES (?, ?)",
            id_val, desc
        )
    #Hacemos el commit o "guardamos"...
    conn.commit()
```
